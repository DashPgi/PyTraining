```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = np.random.default_rng().integers(low=1, high=100, size=100)
Index = np.random.default_rng().integers(low=1, high=100, size=100)
height = np.random.default_rng().integers(low=1, high=100, size=100)



line3d = plt.subplot(projection="3d")


line3d.plot(data,Index,height)
```
